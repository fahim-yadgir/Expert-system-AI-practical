```python
print("=== Disease Diagnosis Expert System ===")

fever = input("Do you have fever? (yes/no): ").lower()
cough = input("Do you have cough? (yes/no): ").lower()
cold = input("Do you have cold? (yes/no): ").lower()
headache = input("Do you have headache? (yes/no): ").lower()
stomach_pain = input("Do you have stomach pain? (yes/no): ").lower()

if fever == "yes" and cough == "yes" and cold == "yes":
    print("Possible Diagnosis: Flu")

elif fever == "yes" and headache == "yes":
    print("Possible Diagnosis: Viral Fever")

elif stomach_pain == "yes" and fever == "yes":
    print("Possible Diagnosis: Stomach Infection")

elif cold == "yes" and cough == "yes":
    print("Possible Diagnosis: Common Cold")

else:
    print("Diagnosis could not be determined.")
```
